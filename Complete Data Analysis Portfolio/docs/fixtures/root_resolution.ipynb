# Nested notebook root-resolution fixture

Confirms the documented bootstrap works when the kernel starts in this nested folder.

```python
from pathlib import Path
import sys

def find_project_root(start=Path.cwd()):
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate / 'PRD.md').is_file() and (candidate / 'src').is_dir():
            return candidate
    raise FileNotFoundError('Run this notebook from inside the portfolio project.')

PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
from src.contracts import DOMAINS
assert len(DOMAINS) == 5
assert (PROJECT_ROOT / 'datasets' / 'catalog.csv').is_file()
print('Project root and shared imports resolved.')
```
